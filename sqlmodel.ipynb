```python
from sqlmodel import Field, SQLModel, Relationship, create_engine, Session
import pandas as pd
```

```python
# Metabase/apache Superset for viz
engine = create_engine("sqlite:///database.db")
```

```python
class B(SQLModel):
    p: int = Field(default=0, primary_key=True)
    _df: pd.DataFrame = None

    def model_post_init(self, context):
        self._df = pd.DataFrame()
```

```python
class A(SQLModel, table=True):
    __tablename__ = "a"
    id: int = Field(default=None, primary_key=True)
    name: str = Field(default="")
    b_id: int = Field(default=None, foreign_key="b.p")
    b: B = Relationship(back_populates="a")
    _df: pd.DataFrame = None

    def model_post_init(self, context):
        self._df = pd.DataFrame()
```

```python
class A(SQLModel, table=True):
    id: str = Field(default="Yes", primary_key=True)
    fi: int = 5
```

```python
SQLModel.metadata.create_all(engine)
with Session(engine) as session:
    session.add(A())
    session.commit()
```
