```python
import sys

sys.path.append("../../..")
from setup_figs import clean_names, pd, sns, plt
```

```python
hue_order = [
    "Subj. num.",
    "Attachment site",
    "Relative clause type",
    "Obj. num.",
    "Verb freq.",
    "(Obj. freq. x Embed. freq.)",
    "Embed. num.",
]
```

```python
fi = clean_names(pd.read_parquet("0.parquet"))
fi = fi[fi.Feature.isin(hue_order)]
```

```python
plt.figure(figsize=(5, 3))
ax = sns.lineplot(
    fi,
    x="representations.layer",
    y="mean",
    hue="Feature",
    hue_order=hue_order,
    style="Feature",
    markers=True,
    dashes=False,
    palette="tab10",
    legend=None,
)

for feature, color in zip(hue_order, sns.color_palette("tab10")):
    group = fi[fi.Feature == feature]
    ax.fill_between(
        group["representations.layer"],
        group["mean"] - group["std"],
        group["mean"] + group["std"],
        alpha=0.2,
        step=None,
        color=color,
    )

ax.set_xlabel("Layer")
ax.set_ylabel("Baseline FI")
ax.set_xticks([1, 3, 6, 9, 12])
sns.despine(trim=True)
# ax.legend(loc="center left", bbox_to_anchor=(1, 0.5), ncol=2, title="Feature")
plt.savefig(
    "../../../paper/figs/BERT_RC_mean/baselines/encoding.pdf", bbox_inches="tight"
)
plt.show()
```
