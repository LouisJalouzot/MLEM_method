```python
import sys

sys.path.append("../../..")
from setup_figs import pd, sns, plt
```

```python
spearman = pd.read_parquet("1.parquet")
spearman = spearman[spearman.split == "test"]
```

```python
plt.Figure(figsize=(5, 3))
uni = (
    spearman.groupby(["trainer.representations.layer", "trainer.representations.units"])[
        "mean"
    ]
    .mean()
    .reset_index()
)
uni["Mode"] = "Univariate"
ax = sns.boxplot(
    uni,
    x="trainer.representations.layer",
    y="mean",
    hue="Mode",
    whis=3,
    palette=sns.color_palette("tab10")[0:],
)
multi = spearman[spearman["trainer.representations.units"].isnull()].copy()
multi["trainer.representations.layer"] -= 1
multi["Mode"] = "Multivariate"
sns.lineplot(
    multi,
    x="trainer.representations.layer",
    y="mean",
    hue="Mode",
    palette=sns.color_palette("tab10")[1:],
    marker="o",
    markersize=10,
)
ax.set_ylim(-0.05, 1)
ax.set_ylabel(r"Test Spearman $\rho$")
ax.set_xlabel("Layer")
sns.move_legend(ax, loc="center left", bbox_to_anchor=(1, 0.5), title=None)
sns.despine(trim=True)
plt.savefig(
    "../../../paper/figs/BERT_RC_mean/univariate/spearman.pdf", bbox_inches="tight"
)
plt.show()
```
